### Celda 3: Función para localizar la imagen en los directorios de imágenes


```python
def find_image_path(img_name):
    # Recorre todos los directorios que comienzan con "images_"
    for folder in os.listdir():
        if folder.startswith("images_") and os.path.isdir(folder):
            path = os.path.join(folder, img_name)
            if os.path.exists(path):
                return path
    return None

# Ejemplo de prueba
print("Ruta de la primera imagen de train_files:", find_image_path(train_files[0]))
```
