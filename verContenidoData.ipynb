```python
import os

# Encontrar el directorio de datos
def find_data_directory():
    current = os.getcwd()
    # Si estamos en el directorio data
    if any(d.startswith("images_") for d in os.listdir()):
        return current
    # Si data está en un subdirectorio
    for d in os.listdir():
        if os.path.isdir(d):
            os.chdir(d)
            if any(sd.startswith("images_") for sd in os.listdir()):
                return os.getcwd()
            os.chdir("..")
    raise Exception("No se encontró el directorio de datos")

# Cambiar al directorio correcto
os.chdir(find_data_directory())
print("Directorio de trabajo:", os.getcwd())
```

```python
print("Contenido del directorio actual:")
for item in os.listdir():
    print(f"- {item}")

# Verificar específicamente las carpetas de imágenes
image_folders = [f for f in os.listdir() if f.startswith("images_")]
print("\nCarpetas de imágenes encontradas:")
print(image_folders)
```
